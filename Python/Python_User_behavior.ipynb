##**Introduction**

This project focuses on analyzing user behavior data using Python. The dataset contains detailed information about users, including age, gender, device model, operating system, app usage time, screen-on time, battery consumption, and data usage.

The main objective of this project is to understand how users interact with their devices and identify key patterns in usage behavior. By performing data cleaning, exploratory data analysis (EDA), and visualization, the project aims to uncover relationships between different factors such as device type, operating system, and user engagement levels.

This analysis helps in gaining insights into user preferences, identifying high-usage patterns, and understanding how different variables impact overall user behavior. The findings from this project can be useful for improving user experience, optimizing application performance, and supporting data-driven decision-making.

```python
# Task 1: Import Libraries & Load Data

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
file_path = r"/content/Python_User Behavior.csv"
df = pd.read_csv(file_path)

# Display first few rows
print("First 5 rows of the dataset:")
print(df.head())
```

```python
# Task 2: Dataset Overview

print("\nShape of the dataset:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset Info:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
# Task 3: Data Cleaning

# Check missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Check duplicates
print("\nNumber of duplicate rows:", df.duplicated().sum())

# Drop duplicates
df = df.drop_duplicates()

# Clean column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

print("\nCleaned column names:")
print(df.columns.tolist())
```

```python
# EXPLORATORY DATA ANALYSIS

# Task 4: Univariate Analysis

# Age Distribution
plt.figure()
sns.histplot(df['age'], bins=10, kde=True)
plt.title("Age Distribution")
plt.show()

# Gender Distribution
plt.figure()
sns.countplot(x=df['gender'])
plt.title("Gender Distribution")
plt.show()

# Device Model Distribution
plt.figure()
sns.countplot(x=df['device_model'])
plt.xticks(rotation=45)
plt.title("Device Model Distribution")
plt.show()
```

```python
# Task 5: Bivariate Analysis

# Usage Time vs Device
plt.figure()
sns.barplot(x='device_model', y='app_usage_time_(min/day)', data=df)
plt.xticks(rotation=45)
plt.title("Usage Time by Device Model")
plt.show()

# Usage Time vs OS
plt.figure()
sns.barplot(x='operating_system', y='app_usage_time_(min/day)', data=df)
plt.title("Usage Time by Operating System")
plt.show()

# Screen Time vs Age
plt.figure()
sns.lineplot(x='age', y='screen_on_time_(hours/day)', data=df)
plt.title("Screen Time by Age")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Task 6: Multivariate Analysis

numeric_cols = df.select_dtypes(include=np.number).columns

if len(numeric_cols) >= 3:
    sns.pairplot(df[numeric_cols[:4]])
    plt.suptitle("Pairplot of Key Numerical Variables", y=1.02)
    plt.show()
```

```python
# Task 7: Data Distribution & Outliers

numeric_cols = df.select_dtypes(include=np.number).columns

plt.figure(figsize=(10,5))
sns.boxplot(data=df[numeric_cols])
plt.title("Outlier Detection (Boxplot)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 8: Visualization Summary

# Distribution of Usage Time
plt.figure()
sns.histplot(df['app_usage_time_(min/day)'], kde=True)
plt.title("App Usage Time Distribution")
plt.show()

# OS Count
plt.figure()
sns.countplot(x=df['operating_system'])
plt.title("Operating System Distribution")
plt.show()

# Heatmap Summary
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="viridis")
plt.title("Correlation Heatmap (Summary)")=
plt.show()
```

```python
# ANALYSIS

# High usage users
high_usage = df[df['app_usage_time_(min/day)'] > 300]
print("Number of High Usage Users:", len(high_usage))

# Average usage by gender
print(df.groupby('gender')['app_usage_time_(min/day)'].mean())
```

```python
# Task 9: Key Insights

print("\nKey Insights:")
print("1. Certain device models show higher app usage compared to others.")
print("2. Users of different operating systems exhibit varying usage patterns.")
print("3. Screen time generally varies across different age groups.")
print("4. High app usage leads to increased battery drain and data consumption.")
print("5. Some users fall into a high usage category, indicating strong engagement.")
```
